### 2. Library imports

````python
import pandas as pd
import seaborn as sns
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import matplotlib_inline.backend_inline
from sklearn.decomposition import PCA

matplotlib_inline.backend_inline.set_matplotlib_formats('svg')
````

### 3. Data read and basic EDA

````python
# Load dataset
rock = pd.read_csv('rock.csv')
rock.head()
````

````python
# brief stats
print(rock.shape)
rock.describe()
````

````python
# Visual exploration. Display correlation matrix using a heatmap
plt.figure(figsize=(4, 4))
sns.heatmap(rock.corr(), cmap="YlGnBu", annot=True);
````

````python
# Visual exploration. Scatter plot using perimeter and area
plt.figure(figsize=(4, 4))
plt.scatter(rock['Perimeter'], rock['Area'])
plt.xlabel('Perimeter', fontsize=14);
plt.ylabel('Area', fontsize=14);
````

````python
# Visual exploration. Scatter plot + linear regression line
model = st.linregress(rock['Perimeter'], rock['Area'])
plt.figure(figsize=(4, 4))
plt.scatter(rock['Perimeter'], rock['Area'])
x = np.linspace(0, 5000, 10000)
y = model[0] * x + model[1]
plt.plot(x, y, '-r', linewidth=2.5)
plt.xlabel('Perimeter', fontsize=14);
plt.ylabel('Area', fontsize=14);
````

### 4. Data pre-processing (standar scaler) 

````python
scaler = StandardScaler()
rock = pd.DataFrame(
    scaler.fit_transform(rock), columns=['Area', 'Perimeter', 'Shape', 'Permeability']
)
````

### 5. Model initialization

````python
# number of components equal to the number of input features
pcaModel = PCA(n_components=4);
````

### 6. Model training

````python
pcaModel.fit(rock)
````

### 7. Model predictions

- display model output
- show explained variance
- compute factor loadings

````python
# Display the components
pcaModel.components_
````

````python
# Display the explained variance (eigenvalues)
pcaModel.explained_variance_
````

````python
# Proportion of Variance Explained (PVE)
pcaModel.explained_variance_ratio_
````

````python
# Show the factor loadings
pcaModel.components_.T * np.sqrt(pcaModel.explained_variance_)
````

### 8. Scree plots

````python
# Scree plot with eigenvalues on the vertical axis
xint = range(0, 5)
plt.xticks(xint)
plt.plot([1, 2, 3, 4], pcaModel.explained_variance_, 'b*-')
plt.xlabel('Factors', fontsize='14');
plt.ylabel('Eigenvalues', fontsize='14');
````

````python
# Scree plot with PVE on the vertical axis
xint = range(0, 5)
plt.xticks(xint)
plt.plot([1, 2, 3, 4], pcaModel.explained_variance_ratio_, 'b*-')
plt.xlabel('Factors', fontsize='14');
plt.ylabel('PVE', fontsize='14');
````

### 9. PCA and eigen decomposition

````python
# creating the dataset matrix 
X = np.array(rock[['Area', 'Perimeter', 'Shape', 'Permeability']])
````

````python
print(f"Dimensions of X = {X.shape}")
n = X.shape[0] # number of datapoints
````

````python
XTX = np.dot(X.T,X)
````

````python
eigenvalues, eigenvectors = np.linalg.eig(XTX)
````

Compare ```eigenvalues/(n-1)``` to ```pcaModel.explained_variance_```.

````python
print(f"Scaled eigenvalues (eigen decomposition) = {eigenvalues/(n-1)}")
print(f"pcaModel.explained_variance_             = {pcaModel.explained_variance_}")
````
